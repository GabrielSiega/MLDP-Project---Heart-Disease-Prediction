```python
import numpy as np
import pandas as pd
import joblib
from flask import Flask, request, jsonify, render_template

# Initialize Flask app
app = Flask(__name__)

# Load the trained model
model = joblib.load('Models/trained_HD_model.pkl')

# Home route (optional HTML form)
@app.route('/')
def home():
    return render_template('index.html')

# Prediction API route
@app.route('/predict', methods=['POST'])
def predict():
    # Expect JSON input or form data
    if request.is_json:
        data = request.get_json()
    else:
        data = request.form.to_dict()

    # Convert input to DataFrame (keys must match training features)
    input_df = pd.DataFrame([data])

    # Predict using the model
    prediction = model.predict(input_df)[0]

    # Return result
    return jsonify({
        "prediction": str(prediction),
        "message": f"Predicted severity of heart disease presence: {prediction}"
    })

if __name__ == "__main__":
    app.run(debug=True)
```
